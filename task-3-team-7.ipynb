```python
# ============================================================
# TASK 3: CUSTOM CNN FROM SCRATCH
# Brain Tumor MRI — 44 Classes
# Same split, same augmentation, same optimizer as Task 2
# ============================================================


# ============================================================
# SECTION 1: ENVIRONMENT SETUP AND DEPENDENCIES
# ============================================================
# (Already imported in Task 2 — include these if running standalone)

import os, time, random, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from collections import Counter

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, Subset
from torchvision import transforms
from torchvision.datasets import ImageFolder

from sklearn.model_selection import StratifiedKFold, train_test_split
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    classification_report, confusion_matrix,
    roc_curve, auc
)
from sklearn.preprocessing import label_binarize
from scipy.stats import ttest_rel

warnings.filterwarnings("ignore")

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark     = False

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"✅ Device : {DEVICE}")


# ============================================================
# SECTION 2: DATA LOADING AND PREPROCESSING
# (Identical split, augmentation, and loaders as Task 2)
# ============================================================

IMG_SIZE    = 224
BATCH_SIZE  = 32
LR          = 1e-3
N_FOLDS     = 5
NUM_WORKERS = 4

# Early stopping settings
ES_PATIENCE  = 5
ES_MIN_EPOCH = 5          # never stop before epoch 5

DATA_DIR = Path("/kaggle/input/datasets/fernando2rad/brain-tumor-mri-images-44c")

# ── Same transforms as Task 2 ──
train_transforms = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(15),
    transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225]),
])

val_test_transforms = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225]),
])

# ── Load full dataset ──
full_dataset = ImageFolder(root=DATA_DIR)
CLASS_NAMES  = full_dataset.classes
NUM_CLASSES  = len(CLASS_NAMES)
all_labels   = np.array([label for _, label in full_dataset.samples])
all_indices  = np.arange(len(full_dataset))

print(f"✅ Total images    : {len(full_dataset)}")
print(f"✅ Number of classes: {NUM_CLASSES}")

# ── Exact same 70/15/15 stratified split as Task 2 ──
train_idx, temp_idx, train_labels, temp_labels = train_test_split(
    all_indices, all_labels,
    test_size=0.30,
    stratify=all_labels,
    random_state=SEED
)
val_idx, test_idx, _, _ = train_test_split(
    temp_idx, temp_labels,
    test_size=0.50,
    stratify=temp_labels,
    random_state=SEED
)

print(f"✅ Train : {len(train_idx)} | Val : {len(val_idx)} | Test : {len(test_idx)}")

# ── Build subsets ──
def make_subset(indices, transform):
    ds = ImageFolder(root=DATA_DIR, transform=transform)
    return Subset(ds, indices)

train_dataset = make_subset(train_idx, train_transforms)
val_dataset   = make_subset(val_idx,   val_test_transforms)
test_dataset  = make_subset(test_idx,  val_test_transforms)

val_loader  = DataLoader(val_dataset,  batch_size=BATCH_SIZE,
                         shuffle=False, num_workers=NUM_WORKERS,pin_memory=True,persistent_workers=True,prefetch_factor=2)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE,
                         shuffle=False, num_workers=NUM_WORKERS,
                           pin_memory=True,persistent_workers=True,prefetch_factor=2)

# ── Same 5-Fold splits on training portion as Task 2 ──
skf = StratifiedKFold(n_splits=N_FOLDS, shuffle=True, random_state=SEED)
train_labels_for_fold = all_labels[train_idx]
fold_splits = list(skf.split(train_idx, train_labels_for_fold))

print(f"✅ Fold splits ready: {N_FOLDS} folds")


# ============================================================
# SECTION 3: CUSTOM CNN ARCHITECTURE
# Conv → BN → ReLU → Pool (5 blocks) → Dropout → Dense → Softmax(44)
# ============================================================

class ConvBlock(nn.Module):
    """Single Conv → BN → ReLU → MaxPool block."""
    def __init__(self, in_channels, out_channels,
                 kernel_size=3, pool=True):
        super().__init__()
        layers = [
            nn.Conv2d(in_channels, out_channels,
                      kernel_size=kernel_size, padding=kernel_size // 2,
                      bias=False),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(inplace=True),
        ]
        if pool:
            layers.append(nn.MaxPool2d(kernel_size=2, stride=2))
        self.block = nn.Sequential(*layers)

    def forward(self, x):
        return self.block(x)


class CustomCNN(nn.Module):
    """
    5-block Custom CNN built from scratch.

    Input  : (B, 3, 224, 224)
    Block 1: Conv(3→32)   → BN → ReLU → Pool  → (B, 32,  112, 112)
    Block 2: Conv(32→64)  → BN → ReLU → Pool  → (B, 64,   56,  56)
    Block 3: Conv(64→128) → BN → ReLU → Pool  → (B, 128,  28,  28)
    Block 4: Conv(128→256)→ BN → ReLU → Pool  → (B, 256,  14,  14)
    Block 5: Conv(256→512)→ BN → ReLU → Pool  → (B, 512,   7,   7)
    GAP    : (B, 512, 1, 1) → Flatten → (B, 512)
    Head   : Dropout(0.5) → FC(512→256) → ReLU
             → Dropout(0.4) → FC(256→44) → Softmax
    """
    def __init__(self, num_classes=NUM_CLASSES,
                 dropout1=0.5, dropout2=0.4):
        super().__init__()

        self.features = nn.Sequential(
            ConvBlock(3,   32,  kernel_size=3, pool=True),   # Block 1
            ConvBlock(32,  64,  kernel_size=3, pool=True),   # Block 2
            ConvBlock(64,  128, kernel_size=3, pool=True),   # Block 3
            ConvBlock(128, 256, kernel_size=3, pool=True),   # Block 4
            ConvBlock(256, 512, kernel_size=3, pool=True),   # Block 5
        )

        # Global Average Pooling — makes the model resolution-agnostic
        self.gap = nn.AdaptiveAvgPool2d(1)

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Dropout(p=dropout1),
            nn.Linear(512, 256),
            nn.ReLU(inplace=True),
            nn.Dropout(p=dropout2),
            nn.Linear(256, num_classes),
            # NOTE: No Softmax here — nn.CrossEntropyLoss applies
            # log-softmax internally. Softmax applied only at inference.
        )

    def forward(self, x):
        x = self.features(x)
        x = self.gap(x)
        x = self.classifier(x)
        return x


def build_custom_cnn():
    print("\n── Building Custom CNN (5 Conv blocks, scratch) ──")
    model = CustomCNN(num_classes=NUM_CLASSES, dropout1=0.5, dropout2=0.4)
    total_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"   Trainable parameters: {total_params:,}")
    return model

# ── Quick architecture sanity check ──
_dummy = build_custom_cnn().to(DEVICE)
_x     = torch.randn(2, 3, 224, 224).to(DEVICE)
_out   = _dummy(_x)
print(f"   Output shape: {_out.shape}  ✅")
del _dummy, _x, _out
torch.cuda.empty_cache()


# ============================================================
# SECTION 4: TRAINING WITH TIME TRACKING + EARLY STOPPING
# ============================================================

def set_seed(seed=SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)


def train_one_epoch(model, loader, criterion, optimizer):
    """One full pass over the training loader."""
    model.train()
    running_loss, correct, total = 0.0, 0, 0
    for inputs, labels in loader:
        inputs, labels = inputs.to(DEVICE), labels.to(DEVICE)
        optimizer.zero_grad()
        outputs = model(inputs)
        loss    = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * inputs.size(0)
        _, preds = torch.max(outputs, 1)
        correct  += (preds == labels).sum().item()
        total    += labels.size(0)
    return running_loss / total, correct / total


def evaluate(model, loader):
    """Evaluate model — returns loss, acc, preds, labels, probs."""
    model.eval()
    criterion = nn.CrossEntropyLoss()
    running_loss, correct, total = 0.0, 0, 0
    all_preds, all_labels, all_probs = [], [], []
    with torch.no_grad():
        for inputs, labels in loader:
            inputs, labels = inputs.to(DEVICE), labels.to(DEVICE)
            outputs = model(inputs)
            loss    = criterion(outputs, labels)
            running_loss += loss.item() * inputs.size(0)
            probs = torch.softmax(outputs, dim=1)
            _, preds = torch.max(outputs, 1)
            correct  += (preds == labels).sum().item()
            total    += labels.size(0)
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())
            all_probs.extend(probs.cpu().numpy())
    return (running_loss / total, correct / total,
            np.array(all_preds), np.array(all_labels), np.array(all_probs))


class EarlyStopping:
    """
    Stops training when val_loss stops improving.
    - Waits at least `min_epoch` epochs before activating.
    - Saves best model weights automatically.
    """
    def __init__(self, patience=ES_PATIENCE, min_epoch=ES_MIN_EPOCH,
                 verbose=True):
        self.patience   = patience
        self.min_epoch  = min_epoch
        self.verbose    = verbose
        self.best_loss  = np.inf
        self.counter    = 0
        self.best_state = None
        self.triggered  = False

    def step(self, val_loss, model, epoch):
        if epoch < self.min_epoch:
            # Always save during warm-up phase
            self.best_state = {k: v.cpu().clone()
                               for k, v in model.state_dict().items()}
            return False   # do not stop

        if val_loss < self.best_loss:
            self.best_loss  = val_loss
            self.counter    = 0
            self.best_state = {k: v.cpu().clone()
                               for k, v in model.state_dict().items()}
            if self.verbose:
                print(f"    ↓ Val loss improved → {val_loss:.4f}. "
                      f"Best weights saved.")
        else:
            self.counter += 1
            if self.verbose:
                print(f"    ✗ No improvement. "
                      f"EarlyStopping counter: {self.counter}/{self.patience}")
            if self.counter >= self.patience:
                self.triggered = True
                if self.verbose:
                    print(f"    ⏹  Early stopping triggered at epoch {epoch}.")
                return True   # stop training
        return False


def train_custom_cnn(model, train_loader, val_loader_ref,
                     max_epochs=50, fold=None, tag="CustomCNN"):
    """
    Full training loop with:
      - Adam optimizer (same LR as Task 2)
      - ReduceLROnPlateau scheduler
      - Early stopping (patience=5, min 5 epochs)
      - Time tracking per epoch
    """
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=LR, weight_decay=1e-4)
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(
        optimizer, mode='min', factor=0.5, patience=3
    )
    es = EarlyStopping(patience=ES_PATIENCE, min_epoch=ES_MIN_EPOCH)

    history    = {"train_loss": [], "train_acc": [],
                  "val_loss":   [], "val_acc":   []}
    total_time = 0.0
    label      = f"[{tag}{'  Fold ' + str(fold) if fold else ''}]"

    print(f"\n{label} Training started — max {max_epochs} epochs "
          f"| LR={LR} | ES patience={ES_PATIENCE} | min_epoch={ES_MIN_EPOCH}")

    for epoch in range(1, max_epochs + 1):
        t0 = time.time()
        tr_loss, tr_acc = train_one_epoch(model, train_loader,
                                           criterion, optimizer)
        vl_loss, vl_acc, _, _, _ = evaluate(model, val_loader_ref)
        scheduler.step(vl_loss)
        elapsed     = time.time() - t0
        total_time += elapsed

        history["train_loss"].append(tr_loss)
        history["train_acc"].append(tr_acc)
        history["val_loss"].append(vl_loss)
        history["val_acc"].append(vl_acc)

        print(f"  Epoch {epoch:03d}/{max_epochs} | "
              f"Train Loss: {tr_loss:.4f} Acc: {tr_acc:.4f} | "
              f"Val Loss: {vl_loss:.4f} Acc: {vl_acc:.4f} | "
              f"Time: {elapsed:.1f}s")

        if es.step(vl_loss, model, epoch):
            break   # early stopping triggered

    # Restore best weights
    if es.best_state is not None:
        model.load_state_dict(
            {k: v.to(DEVICE) for k, v in es.best_state.items()}
        )
        stopped_at = len(history["train_loss"])
        print(f"\n{label} Restored best weights "
              f"(best val_loss={es.best_loss:.4f}, "
              f"stopped at epoch {stopped_at})")

    print(f"{label} Total training time: "
          f"{total_time:.1f}s ({total_time/60:.1f} min)")
    return history, total_time


# ============================================================
# SECTION 5: PLOT TRAINING AND VALIDATION CURVES
# ============================================================

def plot_curves(history, model_name="CustomCNN", fold=None):
    tag    = f"{model_name}" + (f" — Fold {fold}" if fold else " — Final")
    epochs = range(1, len(history["train_loss"]) + 1)

    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    fig.suptitle(f"Training & Validation Curves — {tag}",
                 fontsize=14, fontweight='bold')

    axes[0].plot(epochs, history["train_loss"],
                 label="Train Loss", color="royalblue")
    axes[0].plot(epochs, history["val_loss"],
                 label="Val Loss",   color="tomato")
    axes[0].set_title("Loss")
    axes[0].set_xlabel("Epoch")
    axes[0].set_ylabel("Loss")
    axes[0].legend()

    axes[1].plot(epochs, history["train_acc"],
                 label="Train Acc", color="royalblue")
    axes[1].plot(epochs, history["val_acc"],
                 label="Val Acc",   color="tomato")
    axes[1].set_title("Accuracy")
    axes[1].set_xlabel("Epoch")
    axes[1].set_ylabel("Accuracy")
    axes[1].legend()

    plt.tight_layout()
    fname = f"curves_customcnn_fold{fold if fold else 'final'}.png"
    plt.savefig(fname, dpi=150)
    plt.show()
    print(f"  📊 Curve saved: {fname}")


# ============================================================
# SECTION 6 & 7: EVALUATION AND PERFORMANCE METRICS
# ============================================================

def report_metrics(y_true, y_pred, y_probs,
                   model_name="CustomCNN", training_time=0):
    acc  = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred,
                           average='macro', zero_division=0)
    rec  = recall_score(y_true, y_pred,
                        average='macro', zero_division=0)
    f1   = f1_score(y_true, y_pred,
                    average='macro', zero_division=0)

    print(f"\n{'='*60}")
    print(f"  PERFORMANCE METRICS — {model_name}")
    print(f"{'='*60}")
    print(f"  Accuracy          : {acc:.4f}")
    print(f"  Precision (macro) : {prec:.4f}")
    print(f"  Recall    (macro) : {rec:.4f}")
    print(f"  F1-Score  (macro) : {f1:.4f}")
    print(f"  Training Time     : {training_time:.1f}s "
          f"({training_time/60:.1f} min)")
    return acc, prec, rec, f1


# ============================================================
# SECTION 8: CLASSIFICATION REPORT
# Class accuracy, precision, recall, f1-score for all 44 classes
# ============================================================

def print_classification_report(y_true, y_pred,
                                  model_name="CustomCNN"):
    print(f"\n{'='*60}")
    print(f"  CLASSIFICATION REPORT — {model_name}")
    print(f"{'='*60}")
    print(classification_report(
        y_true, y_pred,
        target_names=CLASS_NAMES,
        digits=4
    ))


# ============================================================
# SECTION 9: CONFUSION MATRIX VISUALIZATION
# ============================================================

def plot_confusion_matrix(y_true, y_pred, model_name="CustomCNN"):
    cm = confusion_matrix(y_true, y_pred)
    fig, ax = plt.subplots(figsize=(28, 26))
    ax.set_title(f"Confusion Matrix — {model_name}",
                 fontsize=16, fontweight='bold')
    sns.heatmap(cm, annot=False, fmt='d', cmap='Blues',
                xticklabels=CLASS_NAMES,
                yticklabels=CLASS_NAMES, ax=ax)
    ax.set_xlabel("Predicted Label", fontsize=12)
    ax.set_ylabel("True Label",      fontsize=12)
    plt.xticks(rotation=90, fontsize=6)
    plt.yticks(rotation=0,  fontsize=6)
    plt.tight_layout()
    fname = f"confusion_matrix_{model_name.lower().replace(' ', '_')}.png"
    plt.savefig(fname, dpi=150)
    plt.show()
    print(f"  📊 Confusion matrix saved: {fname}")


# ============================================================
# SECTION 10: ROC CURVE AND AUC (ONE-VS-REST)
# ============================================================

def plot_roc_curves(y_true, y_probs, model_name="CustomCNN"):
    y_bin = label_binarize(y_true, classes=list(range(NUM_CLASSES)))
    fpr, tpr, roc_auc_dict = {}, {}, {}

    for i in range(NUM_CLASSES):
        fpr[i], tpr[i], _ = roc_curve(y_bin[:, i], y_probs[:, i])
        roc_auc_dict[i]   = auc(fpr[i], tpr[i])

    # Macro-average ROC
    all_fpr  = np.unique(np.concatenate([fpr[i] for i in range(NUM_CLASSES)]))
    mean_tpr = np.zeros_like(all_fpr)
    for i in range(NUM_CLASSES):
        mean_tpr += np.interp(all_fpr, fpr[i], tpr[i])
    mean_tpr  /= NUM_CLASSES
    macro_auc  = auc(all_fpr, mean_tpr)

    plt.figure(figsize=(10, 8))
    plt.title(f"ROC Curves (One-vs-Rest) — {model_name}",
              fontsize=14, fontweight='bold')
    for i in range(NUM_CLASSES):
        plt.plot(fpr[i], tpr[i], alpha=0.25, linewidth=0.8)
    plt.plot(all_fpr, mean_tpr, color='darkred', linewidth=2.5,
             label=f"Macro Avg AUC = {macro_auc:.4f}")
    plt.plot([0, 1], [0, 1], 'k--', linewidth=1)
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.legend(loc="lower right")
    plt.tight_layout()
    fname = f"roc_{model_name.lower().replace(' ', '_')}.png"
    plt.savefig(fname, dpi=150)
    plt.show()
    print(f"  📊 ROC curve saved: {fname}")
    return macro_auc


# ============================================================
# SECTION 11 (prep): STORE FOLD ACCURACIES FOR PAIRED T-TEST
# ============================================================

custom_cnn_fold_accuracies = []
custom_cnn_fold_histories  = []


# ============================================================
# SECTION 4 (execution): 5-FOLD K-FOLD TRAINING LOOP
# ============================================================

print(f"\n{'#'*70}")
print(f"#  TASK 3 — CUSTOM CNN — 5-FOLD STRATIFIED K-FOLD TRAINING")
print(f"{'#'*70}")

for fold_idx, (fold_train_rel, fold_val_rel) in enumerate(fold_splits, 1):

    print(f"\n{'='*60}")
    print(f"  FOLD {fold_idx} / {N_FOLDS} — CUSTOM CNN")
    print(f"{'='*60}")
    set_seed(SEED)

    # Map relative → absolute indices
    fold_train_abs = train_idx[fold_train_rel]
    fold_val_abs   = train_idx[fold_val_rel]

    fold_train_ds = Subset(
        ImageFolder(root=DATA_DIR, transform=train_transforms),
        fold_train_abs
    )
    fold_val_ds = Subset(
        ImageFolder(root=DATA_DIR, transform=val_test_transforms),
        fold_val_abs
    )

    fold_train_loader = DataLoader(
        fold_train_ds, batch_size=BATCH_SIZE,
        shuffle=True,  num_workers=NUM_WORKERS
    )
    fold_val_loader = DataLoader(
        fold_val_ds,   batch_size=BATCH_SIZE,
        shuffle=False, num_workers=NUM_WORKERS
    )

    model = build_custom_cnn().to(DEVICE)
    if torch.cuda.device_count() > 1:
        model = nn.DataParallel(model)

    history, _ = train_custom_cnn(
        model,
        fold_train_loader,
        fold_val_loader,
        max_epochs=50,
        fold=fold_idx
    )
    custom_cnn_fold_histories.append(history)

    # ── SECTION 5: Fold training curves ──
    print(f"\n  SECTION 5 — Training & Validation Curves [Fold {fold_idx}]")
    plot_curves(history, model_name="CustomCNN", fold=fold_idx)

    # ── SECTION 6 & 7: Fold evaluation ──
    print(f"\n  SECTION 6 & 7 — Fold {fold_idx} Evaluation")
    _, fold_acc, y_pred_fold, y_true_fold, _ = evaluate(model, fold_val_loader)
    custom_cnn_fold_accuracies.append(fold_acc)
    print(f"  Fold {fold_idx} Val Accuracy: {fold_acc:.4f}")

    # ── SECTION 8: Per-fold classification report ──
    print(f"\n  SECTION 8 — Classification Report [Fold {fold_idx}]")
    print_classification_report(y_true_fold, y_pred_fold,
                                 model_name=f"CustomCNN Fold {fold_idx}")

    del model
    torch.cuda.empty_cache()


# ── K-Fold summary ──
print(f"\n{'='*60}")
print(f"  K-FOLD SUMMARY — CUSTOM CNN")
print(f"{'='*60}")
print(f"  Per-fold accuracies : "
      f"{[f'{a:.4f}' for a in custom_cnn_fold_accuracies]}")
print(f"  Mean accuracy       : "
      f"{np.mean(custom_cnn_fold_accuracies):.4f} ± "
      f"{np.std(custom_cnn_fold_accuracies):.4f}")


# ============================================================
# FINAL MODEL: TRAIN ON FULL 70% — EVALUATE ON HELD-OUT 15% TEST
# ============================================================

print(f"\n{'#'*70}")
print(f"#  TASK 3 — FINAL MODEL: FULL TRAINING SET → TEST SET EVALUATION")
print(f"{'#'*70}")
set_seed(SEED)

full_train_loader = DataLoader(
    train_dataset, batch_size=BATCH_SIZE,
    shuffle=True,  num_workers=NUM_WORKERS
)

final_model = build_custom_cnn().to(DEVICE)
if torch.cuda.device_count() > 1:
    final_model = nn.DataParallel(final_model)

final_history, total_time = train_custom_cnn(
    final_model,
    full_train_loader,
    val_loader,
    max_epochs=50,
    fold=None
)

# ── SECTION 5: Final model curves ──
print(f"\n  SECTION 5 — Training & Validation Curves [Final Model]")
plot_curves(final_history, model_name="CustomCNN", fold=None)

# ── SECTION 6 & 7: Final evaluation on test set ──
print(f"\n  SECTION 6 & 7 — FINAL EVALUATION ON TEST SET")
_, test_acc, y_pred, y_true, y_probs = evaluate(final_model, test_loader)
acc, prec, rec, f1 = report_metrics(
    y_true, y_pred, y_probs,
    model_name="CustomCNN", training_time=total_time
)

# ── SECTION 8: Classification report (all 44 classes) ──
print(f"\n  SECTION 8 — CLASSIFICATION REPORT — FINAL MODEL")
print_classification_report(y_true, y_pred, model_name="CustomCNN — Final")

# ── SECTION 9: Confusion matrix ──
print(f"\n  SECTION 9 — CONFUSION MATRIX — FINAL MODEL")
plot_confusion_matrix(y_true, y_pred, model_name="CustomCNN")

# ── SECTION 10: ROC Curve and AUC ──
print(f"\n  SECTION 10 — ROC CURVE AND AUC (ONE-VS-REST) — FINAL MODEL")
macro_auc = plot_roc_curves(y_true, y_probs, model_name="CustomCNN")

# ── SECTION 12: Final AUC score ──
print(f"\n{'='*60}")
print(f"  SECTION 12 — FINAL DETECTION AUC SCORE — CUSTOM CNN")
print(f"{'='*60}")
print(f"  Macro-Average AUC (One-vs-Rest): {macro_auc:.4f}")

# Save final model
torch.save(final_model.state_dict(), "customcnn_final.pth")
print(f"\n  ✅ Model saved: customcnn_final.pth")


# ============================================================
# SECTION 11: PAIRED T-TEST
# Compare CustomCNN fold scores vs Task 2 models
# (Paste Task 2 fold scores here, or load from saved arrays)
# ============================================================

print(f"\n{'='*60}")
print(f"  SECTION 11 — PAIRED T-TEST: CustomCNN vs Task 2 Models")
print(f"{'='*60}")

# ── Load Task 2 fold accuracies ──
# Replace these with actual values from Task 2 results
# e.g.  mobilenet_fold_scores = results_mobilenet["fold_accuracies"]
#        resnet_fold_scores    = results_resnet["fold_accuracies"]
#        efficientnet_fold_scores = results_efficientnet["fold_accuracies"]

# If running in the same notebook as Task 2, just reference directly:
try:
    mobilenet_folds    = results_mobilenet["fold_accuracies"]
    resnet_folds       = results_resnet["fold_accuracies"]
    efficientnet_folds = results_efficientnet["fold_accuracies"]
except NameError:
    print("  ⚠ Task 2 results not found in scope.")
    print("    Paste fold accuracy lists manually below:")
    mobilenet_folds    = [0.0] * N_FOLDS   # ← replace
    resnet_folds       = [0.0] * N_FOLDS   # ← replace
    efficientnet_folds = [0.0] * N_FOLDS   # ← replace

cnn_folds = custom_cnn_fold_accuracies

comparisons = [
    ("CustomCNN", "MobileNetV3",   cnn_folds, mobilenet_folds),
    ("CustomCNN", "ResNet50",      cnn_folds, resnet_folds),
    ("CustomCNN", "EfficientNetV2",cnn_folds, efficientnet_folds),
]

for name_a, name_b, scores_a, scores_b in comparisons:
    t_stat, p_val = ttest_rel(scores_a, scores_b)
    sig = "→ Significant ✅" if p_val < 0.05 else "→ Not significant ❌"
    print(f"  {name_a} vs {name_b:<15}: "
          f"t={t_stat:+.4f}, p={p_val:.4f}  {sig}")


# ============================================================
# FINAL COMPARISON: CustomCNN vs Task 2 Models
# ============================================================

print(f"\n{'='*60}")
print(f"  FINAL SUMMARY — ALL 4 MODELS (Task 2 + Task 3)")
print(f"{'='*60}")

# Build summary — include Task 2 results if available
task3_result = {
    "Model"          : "CustomCNN",
    "Test Accuracy"  : f"{acc:.4f}",
    "Precision"      : f"{prec:.4f}",
    "Recall"         : f"{rec:.4f}",
    "F1-Score"       : f"{f1:.4f}",
    "Macro AUC"      : f"{macro_auc:.4f}",
    "Train Time (s)" : f"{total_time:.1f}",
    "KFold Mean Acc" : f"{np.mean(cnn_folds):.4f}",
    "KFold Std Acc"  : f"{np.std(cnn_folds):.4f}",
}

try:
    all_results = [results_mobilenet, results_resnet, results_efficientnet]
    summary_rows = [{
        "Model"          : r["model_name"].upper(),
        "Test Accuracy"  : f"{r['test_accuracy']:.4f}",
        "Precision"      : f"{r['precision']:.4f}",
        "Recall"         : f"{r['recall']:.4f}",
        "F1-Score"       : f"{r['f1']:.4f}",
        "Macro AUC"      : f"{r['macro_auc']:.4f}",
        "Train Time (s)" : f"{r['training_time_s']:.1f}",
        "KFold Mean Acc" : f"{np.mean(r['fold_accuracies']):.4f}",
        "KFold Std Acc"  : f"{np.std(r['fold_accuracies']):.4f}",
    } for r in all_results]
    summary_rows.append(task3_result)
except NameError:
    summary_rows = [task3_result]

summary_df = pd.DataFrame(summary_rows)
print(summary_df.to_string(index=False))
summary_df.to_csv("all_models_comparison.csv", index=False)
print("\n  ✅ Saved: all_models_comparison.csv")

# ── Bar chart: all 4 models ──
fig, axes = plt.subplots(1, 3, figsize=(20, 6))
fig.suptitle("All Models Comparison — Task 2 + Task 3",
             fontsize=16, fontweight='bold')

model_labels = summary_df["Model"].tolist()
colors = ["steelblue", "darkorange", "seagreen", "crimson"]

for ax, (col, title) in zip(axes, [
    ("Test Accuracy", "Test Accuracy"),
    ("F1-Score",      "Macro F1-Score"),
    ("Macro AUC",     "Macro AUC"),
]):
    vals = summary_df[col].astype(float).tolist()
    bars = ax.bar(model_labels, vals,
                  color=colors[:len(model_labels)], edgecolor='black')
    ax.set_title(title, fontweight='bold')
    ax.set_ylim(0, 1.1)
    ax.tick_params(axis='x', rotation=15)
    for bar, val in zip(bars, vals):
        ax.text(bar.get_x() + bar.get_width() / 2,
                bar.get_height() + 0.01,
                f"{val:.4f}", ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.savefig("all_models_final_comparison.png", dpi=150)
plt.show()
print("\n Task 3 Complete.")
```

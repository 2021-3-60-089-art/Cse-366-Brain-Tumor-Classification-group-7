```python
# ============================================================
# TASK 4: ATTENTION-ENHANCED CNN (CBAM)
# Brain Tumor MRI — 44 Classes
# Base: Task 3 Custom CNN + CBAM after every Conv block
# Same split, same hyperparameters as Task 3
# ============================================================


# ============================================================
# SECTION 1: ENVIRONMENT SETUP AND DEPENDENCIES
# ============================================================




# ── Task 2 & 3 results hardcoded from completed notebook outputs ──
TASK2_DIR = "/kaggle/input/notebooks/sumu456/task2-team-h"   
TASK3_DIR = "/kaggle/input/notebooks/sumu456/task3-team-h"  

results_mobilenet = {
    "fold_accuracies": [0.8006, 0.8166, 0.8341, 0.8022, 0.8341],
    "test_accuracy": 0.8527,
    "precision": 0.8554,
    "recall": 0.7964,
    "f1": 0.8141,
    "macro_auc": 0.9938,
    "training_time_s": 566.3,
}
results_resnet = {
    "fold_accuracies": [0.8740, 0.8931, 0.8628, 0.8517, 0.8756],
    "test_accuracy": 0.8884,
    "precision": 0.8947,
    "recall": 0.8372,
    "f1": 0.8535,
    "macro_auc": 0.9889,
    "training_time_s": 693.5,
}
results_efficientnet = {
    "fold_accuracies": [0.8947, 0.9234, 0.9011, 0.8915, 0.8804],
    "test_accuracy": 0.9211,
    "precision": 0.9146,
    "recall": 0.8889,
    "f1": 0.8965,
    "macro_auc": 0.9974,
    "training_time_s": 755.0,
}

custom_cnn_fold_accuracies = [0.7193, 0.7735, 0.4976, 0.7512, 0.7624]

import os, time, random, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns
from pathlib import Path

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, Subset
from torchvision import transforms
from torchvision.datasets import ImageFolder

from sklearn.model_selection import StratifiedKFold, train_test_split
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    classification_report, confusion_matrix,
    roc_curve, auc
)
from sklearn.preprocessing import label_binarize
from scipy.stats import ttest_rel

warnings.filterwarnings("ignore")

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)



torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark     = False
def seed_worker(worker_id):
    worker_seed = torch.initial_seed() % 2**32
    np.random.seed(worker_seed)
    random.seed(worker_seed)

g = torch.Generator()
g.manual_seed(SEED)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"✅ Device : {DEVICE}")


# ============================================================
# SECTION 2: DATA LOADING AND PREPROCESSING
# Exact same split and augmentation as Task 3
# ============================================================

IMG_SIZE    = 224
BATCH_SIZE  = 128
LR          = 1e-3
N_FOLDS     = 5
NUM_WORKERS = 2
ES_PATIENCE  = 5
ES_MIN_EPOCH = 5

DATA_DIR = Path("/kaggle/input/datasets/fernando2rad/brain-tumor-mri-images-44c")

# ── Same transforms as Task 3 ──
train_transforms = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(15),
    transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225]),
])

val_test_transforms = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225]),
])

# ── Load full dataset ──
full_dataset = ImageFolder(root=DATA_DIR)
CLASS_NAMES  = full_dataset.classes
NUM_CLASSES  = len(CLASS_NAMES)
all_labels   = np.array([label for _, label in full_dataset.samples])
all_indices  = np.arange(len(full_dataset))

print(f"✅ Total images     : {len(full_dataset)}")
print(f"✅ Number of classes: {NUM_CLASSES}")

# ── Exact same 70/15/15 split as Task 3 ──
train_idx, temp_idx, train_labels, temp_labels = train_test_split(
    all_indices, all_labels,
    test_size=0.30, stratify=all_labels, random_state=SEED
)
val_idx, test_idx, _, _ = train_test_split(
    temp_idx, temp_labels,
    test_size=0.50, stratify=temp_labels, random_state=SEED
)

print(f"✅ Train: {len(train_idx)} | Val: {len(val_idx)} | Test: {len(test_idx)}")

def make_subset(indices, transform):
    ds = ImageFolder(root=DATA_DIR, transform=transform)
    return Subset(ds, indices)

train_dataset = make_subset(train_idx, train_transforms)
val_dataset   = make_subset(val_idx,   val_test_transforms)
test_dataset  = make_subset(test_idx,  val_test_transforms)

val_loader  = DataLoader(val_dataset,  batch_size=BATCH_SIZE,
                         shuffle=False, num_workers=NUM_WORKERS)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE,
                         shuffle=False, num_workers=NUM_WORKERS)

# ── Same 5-fold splits as Task 3 ──
skf = StratifiedKFold(n_splits=N_FOLDS, shuffle=True, random_state=SEED)
train_labels_for_fold = all_labels[train_idx]
fold_splits = list(skf.split(train_idx, train_labels_for_fold))

print(f"✅ Fold splits ready : {N_FOLDS} folds")


# ============================================================
# SECTION 3: ATTENTION-ENHANCED CNN ARCHITECTURE
# CBAM = Channel Attention Module + Spatial Attention Module
# Plugged in after every Conv → BN → ReLU → Pool block
# ============================================================

# ── CBAM: Channel Attention Module ──
class ChannelAttention(nn.Module):
    """
    Learns WHICH feature channels matter.
    Uses both AvgPool and MaxPool, passes through a shared MLP,
    adds results and applies Sigmoid gate.

    Input  : (B, C, H, W)
    Output : (B, C, 1, 1)  — channel-wise weights
    """
    def __init__(self, in_channels, reduction_ratio=16):
        super().__init__()
        reduced = max(1, in_channels // reduction_ratio)
        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.max_pool = nn.AdaptiveMaxPool2d(1)
        self.shared_mlp = nn.Sequential(
            nn.Flatten(),
            nn.Linear(in_channels, reduced, bias=False),
            nn.ReLU(inplace=True),
            nn.Linear(reduced, in_channels, bias=False),
        )
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        avg_out = self.shared_mlp(self.avg_pool(x))
        max_out = self.shared_mlp(self.max_pool(x))
        scale   = self.sigmoid(avg_out + max_out)
        # Reshape to (B, C, 1, 1) for broadcasting
        return scale.unsqueeze(-1).unsqueeze(-1)


# ── CBAM: Spatial Attention Module ──
class SpatialAttention(nn.Module):
    """
    Learns WHERE in the spatial map to focus.
    Concatenates channel-wise AvgPool and MaxPool,
    passes through a 7×7 conv, applies Sigmoid gate.

    Input  : (B, C, H, W)
    Output : (B, 1, H, W)  — spatial weights
    """
    def __init__(self, kernel_size=7):
        super().__init__()
        self.conv = nn.Conv2d(
            in_channels=2, out_channels=1,
            kernel_size=kernel_size,
            padding=kernel_size // 2,
            bias=False
        )
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        avg_out = torch.mean(x, dim=1, keepdim=True)   # (B,1,H,W)
        max_out, _ = torch.max(x, dim=1, keepdim=True) # (B,1,H,W)
        concat  = torch.cat([avg_out, max_out], dim=1) # (B,2,H,W)
        scale   = self.sigmoid(self.conv(concat))       # (B,1,H,W)
        return scale


# ── CBAM Block (Channel + Spatial combined) ──
class CBAM(nn.Module):
    """
    Full CBAM block:
      x → Channel Attention → x' → Spatial Attention → x''
    Both attentions are applied multiplicatively (element-wise).

    Paper: Woo et al., CBAM: Convolutional Block Attention Module, ECCV 2018
    """
    def __init__(self, in_channels, reduction_ratio=16, spatial_kernel=7):
        super().__init__()
        self.channel_att = ChannelAttention(in_channels, reduction_ratio)
        self.spatial_att = SpatialAttention(spatial_kernel)

    def forward(self, x):
        x = x * self.channel_att(x)   # refine channels
        x = x * self.spatial_att(x)   # refine spatial locations
        return x


# ── Conv Block with CBAM attached ──
class ConvBlockWithCBAM(nn.Module):
    """
    Conv → BN → ReLU → MaxPool → CBAM

    This is the Task 3 ConvBlock with CBAM plugged in after pooling.
    The only structural change from Task 3.
    """
    def __init__(self, in_channels, out_channels,
                 kernel_size=3, reduction_ratio=16):
        super().__init__()
        self.conv_block = nn.Sequential(
            nn.Conv2d(in_channels, out_channels,
                      kernel_size=kernel_size,
                      padding=kernel_size // 2,
                      bias=False),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(kernel_size=2, stride=2),
        )
        self.cbam = CBAM(out_channels, reduction_ratio=reduction_ratio)

    def forward(self, x):
        x = self.conv_block(x)
        x = self.cbam(x)          # ← attention applied here
        return x


# ── Full Attention-Enhanced CNN ──
class AttentionCNN(nn.Module):
    """
    Identical structure to Task 3 CustomCNN,
    with CBAM after every Conv block.

    Input  : (B, 3, 224, 224)
    Block 1: Conv(32)  → BN → ReLU → Pool → CBAM  →  32 × 112 × 112
    Block 2: Conv(64)  → BN → ReLU → Pool → CBAM  →  64 ×  56 ×  56
    Block 3: Conv(128) → BN → ReLU → Pool → CBAM  → 128 ×  28 ×  28
    Block 4: Conv(256) → BN → ReLU → Pool → CBAM  → 256 ×  14 ×  14
    Block 5: Conv(512) → BN → ReLU → Pool → CBAM  → 512 ×   7 ×   7
    GAP    : → 512
    Head   : Dropout(0.5) → FC(256) → ReLU → Dropout(0.4) → FC(44)
    """
    def __init__(self, num_classes=NUM_CLASSES,
                 dropout1=0.5, dropout2=0.4):
        super().__init__()

        self.features = nn.Sequential(
            ConvBlockWithCBAM(3,   32,  reduction_ratio=4),   # small channels → small ratio
            ConvBlockWithCBAM(32,  64,  reduction_ratio=8),
            ConvBlockWithCBAM(64,  128, reduction_ratio=16),
            ConvBlockWithCBAM(128, 256, reduction_ratio=16),
            ConvBlockWithCBAM(256, 512, reduction_ratio=16),
        )

        self.gap = nn.AdaptiveAvgPool2d(1)

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Dropout(p=dropout1),
            nn.Linear(512, 256),
            nn.ReLU(inplace=True),
            nn.Dropout(p=dropout2),
            nn.Linear(256, num_classes),
        )

    def forward(self, x):
        x = self.features(x)
        x = self.gap(x)
        x = self.classifier(x)
        return x


def build_attention_cnn():
    print("\n── Building Attention-Enhanced CNN (CBAM) ──")
    model = AttentionCNN(num_classes=NUM_CLASSES,
                         dropout1=0.5, dropout2=0.4)
    total  = sum(p.numel() for p in model.parameters())
    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"   Total parameters     : {total:,}")
    print(f"   Trainable parameters : {trainable:,}")
    return model

# ── Sanity check ──
_dummy = build_attention_cnn().to(DEVICE)
_x     = torch.randn(2, 3, 224, 224).to(DEVICE)
_out   = _dummy(_x)
print(f"   Output shape : {_out.shape}  ✅")
del _dummy, _x, _out
torch.cuda.empty_cache()


# ============================================================
# SECTION 4: TRAINING WITH TIME TRACKING + EARLY STOPPING
# Identical training utilities to Task 3
# ============================================================

def set_seed(seed=SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)


def train_one_epoch(model, loader, criterion, optimizer):
    model.train()
    running_loss, correct, total = 0.0, 0, 0
    for inputs, labels in loader:
        inputs, labels = inputs.to(DEVICE), labels.to(DEVICE)
        optimizer.zero_grad()
        outputs = model(inputs)
        loss    = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * inputs.size(0)
        _, preds = torch.max(outputs, 1)
        correct  += (preds == labels).sum().item()
        total    += labels.size(0)
    return running_loss / total, correct / total


def evaluate(model, loader):
    model.eval()
    criterion = nn.CrossEntropyLoss()
    running_loss, correct, total = 0.0, 0, 0
    all_preds, all_labels, all_probs = [], [], []
    with torch.no_grad():
        for inputs, labels in loader:
            inputs, labels = inputs.to(DEVICE), labels.to(DEVICE)
            outputs = model(inputs)
            loss    = criterion(outputs, labels)
            running_loss += loss.item() * inputs.size(0)
            probs = torch.softmax(outputs, dim=1)
            _, preds = torch.max(outputs, 1)
            correct  += (preds == labels).sum().item()
            total    += labels.size(0)
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())
            all_probs.extend(probs.cpu().numpy())
    return (running_loss / total, correct / total,
            np.array(all_preds), np.array(all_labels), np.array(all_probs))


class EarlyStopping:
    def __init__(self, patience=ES_PATIENCE, min_epoch=ES_MIN_EPOCH):
        self.patience   = patience
        self.min_epoch  = min_epoch
        self.best_loss  = np.inf
        self.counter    = 0
        self.best_state = None
        self.triggered  = False

    def step(self, val_loss, model, epoch):
        if epoch < self.min_epoch:
            self.best_state = {k: v.cpu().clone()
                               for k, v in model.state_dict().items()}
            return False
        if val_loss < self.best_loss:
            self.best_loss  = val_loss
            self.counter    = 0
            self.best_state = {k: v.cpu().clone()
                               for k, v in model.state_dict().items()}
        else:
            self.counter += 1
            if self.counter >= self.patience:
                self.triggered = True
                return True       # ← actually stops training now
        return False


def train_attention_cnn(model, train_loader, val_loader_ref,
                        max_epochs=60, fold=None,
                        tag="AttentionCNN"):
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=LR, weight_decay=1e-4)
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(
        optimizer, mode='min', factor=0.5, patience=3)
    es = EarlyStopping(patience=ES_PATIENCE, min_epoch=ES_MIN_EPOCH)

    history    = {"train_loss": [], "train_acc": [],
                  "val_loss":   [], "val_acc":   []}
    total_time = 0.0
    label      = f"[{tag}{'  Fold ' + str(fold) if fold else ''}]"

    print(f"\n{label} Training — max {max_epochs} epochs | "
          f"LR={LR} | ES patience={ES_PATIENCE}")

    for epoch in range(1, max_epochs + 1):
        t0 = time.time()
        tr_loss, tr_acc = train_one_epoch(model, train_loader,
                                           criterion, optimizer)
        vl_loss, vl_acc, _, _, _ = evaluate(model, val_loader_ref)
        scheduler.step(vl_loss)
        elapsed     = time.time() - t0
        total_time += elapsed

        history["train_loss"].append(tr_loss)
        history["train_acc"].append(tr_acc)
        history["val_loss"].append(vl_loss)
        history["val_acc"].append(vl_acc)

        print(f"  Epoch {epoch:03d}/{max_epochs} | "
              f"Train Loss: {tr_loss:.4f} Acc: {tr_acc:.4f} | "
              f"Val Loss: {vl_loss:.4f} Acc: {vl_acc:.4f} | "
              f"Time: {elapsed:.1f}s")

        if es.step(vl_loss, model, epoch):
            break

    if es.best_state is not None:
        model.load_state_dict(
            {k: v.to(DEVICE) for k, v in es.best_state.items()}
        )
        print(f"\n{label} Best weights restored "
              f"(best val_loss={es.best_loss:.4f}, "
              f"stopped at epoch {len(history['train_loss'])})")

    print(f"{label} Total time: {total_time:.1f}s "
          f"({total_time/60:.1f} min)")
    return history, total_time


# ============================================================
# SECTION 5: PLOT TRAINING AND VALIDATION CURVES
# ============================================================

def plot_curves(history, model_name="AttentionCNN", fold=None):
    tag    = f"{model_name}" + (f" — Fold {fold}" if fold else " — Final")
    epochs = range(1, len(history["train_loss"]) + 1)

    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    fig.suptitle(f"Training & Validation Curves — {tag}",
                 fontsize=14, fontweight='bold')

    axes[0].plot(epochs, history["train_loss"],
                 label="Train Loss", color="darkorchid")
    axes[0].plot(epochs, history["val_loss"],
                 label="Val Loss",   color="tomato")
    axes[0].set_title("Loss")
    axes[0].set_xlabel("Epoch")
    axes[0].set_ylabel("Loss")
    axes[0].legend()

    axes[1].plot(epochs, history["train_acc"],
                 label="Train Acc", color="darkorchid")
    axes[1].plot(epochs, history["val_acc"],
                 label="Val Acc",   color="tomato")
    axes[1].set_title("Accuracy")
    axes[1].set_xlabel("Epoch")
    axes[1].set_ylabel("Accuracy")
    axes[1].legend()

    plt.tight_layout()
    fname = f"curves_attentioncnn_fold{fold if fold else 'final'}.png"
    plt.savefig(fname, dpi=150)
    plt.show()
    print(f"  📊 Saved: {fname}")


# ============================================================
# SECTION 6 & 7: EVALUATION AND PERFORMANCE METRICS
# ============================================================

def report_metrics(y_true, y_pred, model_name="AttentionCNN",
                   training_time=0):
    acc  = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred, average='macro', zero_division=0)
    rec  = recall_score(y_true, y_pred,    average='macro', zero_division=0)
    f1   = f1_score(y_true, y_pred,        average='macro', zero_division=0)

    print(f"\n{'='*60}")
    print(f"  PERFORMANCE METRICS — {model_name}")
    print(f"{'='*60}")
    print(f"  Accuracy          : {acc:.4f}")
    print(f"  Precision (macro) : {prec:.4f}")
    print(f"  Recall    (macro) : {rec:.4f}")
    print(f"  F1-Score  (macro) : {f1:.4f}")
    print(f"  Training Time     : {training_time:.1f}s "
          f"({training_time/60:.1f} min)")
    return acc, prec, rec, f1


# ============================================================
# SECTION 8: CLASSIFICATION REPORT
# Per-class accuracy, precision, recall, f1 for all 44 classes
# ============================================================

def print_classification_report(y_true, y_pred,
                                  model_name="AttentionCNN"):
    print(f"\n{'='*60}")
    print(f"  CLASSIFICATION REPORT — {model_name}")
    print(f"{'='*60}")
    print(classification_report(
        y_true, y_pred,
        target_names=CLASS_NAMES,
        digits=4
    ))


# ============================================================
# SECTION 9: CONFUSION MATRIX VISUALIZATION
# + Side-by-side diff vs Task 3 CustomCNN
# ============================================================

def plot_confusion_matrix(y_true, y_pred, model_name="AttentionCNN"):
    cm = confusion_matrix(y_true, y_pred)
    fig, ax = plt.subplots(figsize=(28, 26))
    ax.set_title(f"Confusion Matrix — {model_name}",
                 fontsize=16, fontweight='bold')
    sns.heatmap(cm, annot=False, fmt='d', cmap='Purples',
                xticklabels=CLASS_NAMES,
                yticklabels=CLASS_NAMES, ax=ax)
    ax.set_xlabel("Predicted Label", fontsize=12)
    ax.set_ylabel("True Label",      fontsize=12)
    plt.xticks(rotation=90, fontsize=6)
    plt.yticks(rotation=0,  fontsize=6)
    plt.tight_layout()
    fname = f"confusion_matrix_{model_name.lower().replace(' ', '_')}.png"
    plt.savefig(fname, dpi=150)
    plt.show()
    print(f"  📊 Saved: {fname}")
    return cm


def plot_confusion_matrix_diff(cm_task3, cm_task4,
                                model3_name="CustomCNN",
                                model4_name="AttentionCNN"):
    """
    Side-by-side confusion matrices + difference heatmap.
    Green  = AttentionCNN got MORE right than CustomCNN
    Red    = AttentionCNN got FEWER right (regressions)
    White  = no change
    Shows exactly which tumor classes improved with attention.
    """
    diff = cm_task4.astype(int) - cm_task3.astype(int)

    fig, axes = plt.subplots(1, 3, figsize=(52, 20))
    fig.suptitle(
        f"Confusion Matrix Comparison: {model3_name} vs {model4_name}",
        fontsize=18, fontweight='bold'
    )

    # Task 3
    sns.heatmap(cm_task3, annot=False, fmt='d', cmap='Blues',
                xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES,
                ax=axes[0])
    axes[0].set_title(f"{model3_name}", fontsize=14, fontweight='bold')
    axes[0].set_xlabel("Predicted")
    axes[0].set_ylabel("True")
    axes[0].tick_params(axis='x', rotation=90, labelsize=5)
    axes[0].tick_params(axis='y', rotation=0,  labelsize=5)

    # Task 4
    sns.heatmap(cm_task4, annot=False, fmt='d', cmap='Purples',
                xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES,
                ax=axes[1])
    axes[1].set_title(f"{model4_name}", fontsize=14, fontweight='bold')
    axes[1].set_xlabel("Predicted")
    axes[1].set_ylabel("True")
    axes[1].tick_params(axis='x', rotation=90, labelsize=5)
    axes[1].tick_params(axis='y', rotation=0,  labelsize=5)

    # Difference
    sns.heatmap(diff, annot=False, fmt='d',
                cmap='RdYlGn',     # red=worse, green=better
                center=0,
                xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES,
                ax=axes[2])
    axes[2].set_title(
        f"Difference ({model4_name} − {model3_name})\n"
        f"Green = Improved  |  Red = Regressed",
        fontsize=13, fontweight='bold'
    )
    axes[2].set_xlabel("Predicted")
    axes[2].set_ylabel("True")
    axes[2].tick_params(axis='x', rotation=90, labelsize=5)
    axes[2].tick_params(axis='y', rotation=0,  labelsize=5)

    plt.tight_layout()
    plt.savefig("confusion_matrix_diff_task3_vs_task4.png", dpi=150)
    plt.show()
    print("  📊 Saved: confusion_matrix_diff_task3_vs_task4.png")


def plot_per_class_f1_comparison(y_true_t3, y_pred_t3,
                                  y_true_t4, y_pred_t4):
    """
    Bar chart showing per-class F1 improvement/regression
    from CustomCNN (Task 3) → AttentionCNN (Task 4).
    Highlights the hard-to-classify tumor classes.
    """
    from sklearn.metrics import f1_score

    f1_t3 = f1_score(y_true_t3, y_pred_t3,
                     labels=list(range(NUM_CLASSES)),
                     average=None, zero_division=0)
    f1_t4 = f1_score(y_true_t4, y_pred_t4,
                     labels=list(range(NUM_CLASSES)),
                     average=None, zero_division=0)
    delta = f1_t4 - f1_t3

    # Sort by delta to highlight biggest changes
    sorted_idx = np.argsort(delta)
    sorted_classes = [CLASS_NAMES[i] for i in sorted_idx]
    sorted_delta   = delta[sorted_idx]
    colors = ['seagreen' if d >= 0 else 'tomato' for d in sorted_delta]

    fig, ax = plt.subplots(figsize=(10, 18))
    ax.set_title(
        "Per-Class F1 Change: AttentionCNN − CustomCNN\n"
        "Green = Improved  |  Red = Regressed",
        fontsize=13, fontweight='bold'
    )
    bars = ax.barh(sorted_classes, sorted_delta, color=colors)
    ax.axvline(x=0, color='black', linewidth=1.0)
    ax.set_xlabel("F1 Score Change (AttentionCNN − CustomCNN)")
    for bar, val in zip(bars, sorted_delta):
        ax.text(val + (0.002 if val >= 0 else -0.002),
                bar.get_y() + bar.get_height() / 2,
                f"{val:+.3f}", va='center', ha='left' if val >= 0 else 'right',
                fontsize=7)
    plt.tight_layout()
    plt.savefig("per_class_f1_improvement.png", dpi=150)
    plt.show()
    print("  📊 Saved: per_class_f1_improvement.png")


# ============================================================
# SECTION 10: ROC CURVE AND AUC (ONE-VS-REST)
# ============================================================

def plot_roc_curves(y_true, y_probs, model_name="AttentionCNN"):
    y_bin = label_binarize(y_true, classes=list(range(NUM_CLASSES)))
    fpr, tpr, roc_auc_dict = {}, {}, {}

    for i in range(NUM_CLASSES):
        fpr[i], tpr[i], _ = roc_curve(y_bin[:, i], y_probs[:, i])
        roc_auc_dict[i]   = auc(fpr[i], tpr[i])

    all_fpr  = np.unique(np.concatenate([fpr[i] for i in range(NUM_CLASSES)]))
    mean_tpr = np.zeros_like(all_fpr)
    for i in range(NUM_CLASSES):
        mean_tpr += np.interp(all_fpr, fpr[i], tpr[i])
    mean_tpr  /= NUM_CLASSES
    macro_auc  = auc(all_fpr, mean_tpr)

    plt.figure(figsize=(10, 8))
    plt.title(f"ROC Curves (One-vs-Rest) — {model_name}",
              fontsize=14, fontweight='bold')
    for i in range(NUM_CLASSES):
        plt.plot(fpr[i], tpr[i], alpha=0.25, linewidth=0.8)
    plt.plot(all_fpr, mean_tpr, color='darkorchid', linewidth=2.5,
             label=f"Macro Avg AUC = {macro_auc:.4f}")
    plt.plot([0, 1], [0, 1], 'k--', linewidth=1)
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.legend(loc="lower right")
    plt.tight_layout()
    fname = f"roc_{model_name.lower().replace(' ', '_')}.png"
    plt.savefig(fname, dpi=150)
    plt.show()
    print(f"  📊 Saved: {fname}")
    return macro_auc


def plot_roc_overlay_task3_vs_task4(
        y_true_t3, y_probs_t3,
        y_true_t4, y_probs_t4):
    """Overlay macro-average ROC of Task 3 vs Task 4 on one plot."""
    def macro_roc(y_true, y_probs):
        y_bin = label_binarize(y_true, classes=list(range(NUM_CLASSES)))
        fpr_d, tpr_d = {}, {}
        for i in range(NUM_CLASSES):
            fpr_d[i], tpr_d[i], _ = roc_curve(y_bin[:, i], y_probs[:, i])
        all_fpr  = np.unique(np.concatenate([fpr_d[i] for i in range(NUM_CLASSES)]))
        mean_tpr = np.zeros_like(all_fpr)
        for i in range(NUM_CLASSES):
            mean_tpr += np.interp(all_fpr, fpr_d[i], tpr_d[i])
        mean_tpr /= NUM_CLASSES
        return all_fpr, mean_tpr, auc(all_fpr, mean_tpr)

    fpr3, tpr3, auc3 = macro_roc(y_true_t3, y_probs_t3)
    fpr4, tpr4, auc4 = macro_roc(y_true_t4, y_probs_t4)

    plt.figure(figsize=(9, 7))
    plt.title("Macro-Average ROC: CustomCNN vs AttentionCNN",
              fontsize=14, fontweight='bold')
    plt.plot(fpr3, tpr3, color='steelblue', linewidth=2,
             label=f"CustomCNN    AUC = {auc3:.4f}")
    plt.plot(fpr4, tpr4, color='darkorchid', linewidth=2,
             label=f"AttentionCNN AUC = {auc4:.4f}")
    plt.plot([0, 1], [0, 1], 'k--', linewidth=1)
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.legend(loc="lower right")
    plt.tight_layout()
    plt.savefig("roc_overlay_task3_vs_task4.png", dpi=150)
    plt.show()
    print("  📊 Saved: roc_overlay_task3_vs_task4.png")


# ============================================================
# 5-FOLD K-FOLD TRAINING LOOP — ATTENTION CNN
# ============================================================

print(f"\n{'#'*70}")
print(f"#  TASK 4 — ATTENTION CNN — 5-FOLD STRATIFIED K-FOLD TRAINING")
print(f"{'#'*70}")

attention_fold_accuracies = []
attention_fold_histories  = []

for fold_idx, (fold_train_rel, fold_val_rel) in enumerate(fold_splits, 1):

    print(f"\n{'='*60}")
    print(f"  FOLD {fold_idx} / {N_FOLDS} — ATTENTION CNN (CBAM)")
    print(f"{'='*60}")
    set_seed(SEED)

    fold_train_abs = train_idx[fold_train_rel]
    fold_val_abs   = train_idx[fold_val_rel]

    fold_train_ds = Subset(
        ImageFolder(root=DATA_DIR, transform=train_transforms),
        fold_train_abs
    )
    fold_val_ds = Subset(
        ImageFolder(root=DATA_DIR, transform=val_test_transforms),
        fold_val_abs
    )

    fold_train_loader = DataLoader(fold_train_ds, batch_size=BATCH_SIZE,
                                    shuffle=True,  num_workers=NUM_WORKERS,
    worker_init_fn=seed_worker,
    generator=g,)
    fold_val_loader   = DataLoader(fold_val_ds,   batch_size=BATCH_SIZE,
                                    shuffle=False, num_workers=NUM_WORKERS,worker_init_fn=seed_worker
                                    )

    model = build_attention_cnn().to(DEVICE)

    history, _ = train_attention_cnn(
        model, fold_train_loader, fold_val_loader,
        max_epochs=60, fold=fold_idx
    )
    attention_fold_histories.append(history)

    # ── SECTION 5: Fold curves ──
    print(f"\n  SECTION 5 — Training & Validation Curves [Fold {fold_idx}]")
    plot_curves(history, model_name="AttentionCNN", fold=fold_idx)

    # ── SECTION 6 & 7: Fold metrics ──
    print(f"\n  SECTION 6 & 7 — Fold {fold_idx} Evaluation")
    _, fold_acc, y_pred_fold, y_true_fold, _ = evaluate(model, fold_val_loader)
    attention_fold_accuracies.append(fold_acc)
    print(f"  Fold {fold_idx} Val Accuracy: {fold_acc:.4f}")

    # ── SECTION 8: Per-fold classification report ──
    print(f"\n  SECTION 8 — Classification Report [Fold {fold_idx}]")
    print_classification_report(y_true_fold, y_pred_fold,
                                 model_name=f"AttentionCNN Fold {fold_idx}")

    del model
    torch.cuda.empty_cache()


print(f"\n{'='*60}")
print(f"  K-FOLD SUMMARY — ATTENTION CNN")
print(f"{'='*60}")
print(f"  Per-fold accuracies : "
      f"{[f'{a:.4f}' for a in attention_fold_accuracies]}")
print(f"  Mean accuracy       : "
      f"{np.mean(attention_fold_accuracies):.4f} ± "
      f"{np.std(attention_fold_accuracies):.4f}")


# ============================================================
# FINAL MODEL: TRAIN ON FULL 70% → EVALUATE ON 15% TEST SET
# ============================================================

print(f"\n{'#'*70}")
print(f"#  TASK 4 — FINAL MODEL: FULL TRAIN → TEST SET EVALUATION")
print(f"{'#'*70}")

set_seed(SEED)






full_train_loader = DataLoader(
    train_dataset, batch_size=BATCH_SIZE,
    shuffle=True, num_workers=NUM_WORKERS,
    worker_init_fn=seed_worker,
    generator=g
)

final_model = build_attention_cnn().to(DEVICE)

final_history, total_time = train_attention_cnn(
    final_model, full_train_loader, val_loader,
    max_epochs=60, fold=None
)

# ── SECTION 5: Final curves ──
print(f"\n  SECTION 5 — Training & Validation Curves [Final Model]")
plot_curves(final_history, model_name="AttentionCNN", fold=None)

# ── SECTION 6 & 7: Final metrics ──
print(f"\n  SECTION 6 & 7 — FINAL EVALUATION ON TEST SET")
_, test_acc, y_pred_t4, y_true_t4, y_probs_t4 = evaluate(final_model, test_loader)
acc, prec, rec, f1 = report_metrics(
    y_true_t4, y_pred_t4,
    model_name="AttentionCNN", training_time=total_time
)

# ── SECTION 8: Final classification report ──
print(f"\n  SECTION 8 — CLASSIFICATION REPORT — FINAL MODEL")
print_classification_report(y_true_t4, y_pred_t4,
                              model_name="AttentionCNN — Final")

# ── SECTION 9: Confusion matrix ──
print(f"\n  SECTION 9 — CONFUSION MATRIX — FINAL MODEL")
cm_task4 = plot_confusion_matrix(y_true_t4, y_pred_t4,
                                  model_name="AttentionCNN")

# ── SECTION 10: ROC Curve and AUC ──
print(f"\n  SECTION 10 — ROC CURVE AND AUC (ONE-VS-REST)")
macro_auc = plot_roc_curves(y_true_t4, y_probs_t4,
                             model_name="AttentionCNN")

# ── SECTION 12: Final AUC ──
print(f"\n{'='*60}")
print(f"  SECTION 12 — FINAL DETECTION AUC SCORE — ATTENTION CNN")
print(f"{'='*60}")
print(f"  Macro-Average AUC (One-vs-Rest): {macro_auc:.4f}")

torch.save(final_model.state_dict(), "attentioncnn_final.pth")
print(f"\n  ✅ Model saved: attentioncnn_final.pth")


# ============================================================
# SECTION 11: PAIRED T-TEST — AttentionCNN vs CustomCNN (Task 3)
# ============================================================

print(f"\n{'='*60}")
print(f"  SECTION 11 — PAIRED T-TEST: AttentionCNN vs CustomCNN")
print(f"{'='*60}")

custom_folds = custom_cnn_fold_accuracies  

t_stat, p_val = ttest_rel(attention_fold_accuracies, custom_folds)
sig = "→ Significant ✅" if p_val < 0.05 else "→ Not significant ❌"
print(f"  AttentionCNN vs CustomCNN : "
      f"t={t_stat:+.4f}, p={p_val:.4f}  {sig}")


# ============================================================
# TASK 3 vs TASK 4: DIRECT COMPARISON VISUALIZATIONS
# ============================================================

print(f"\n{'#'*70}")
print(f"#  TASK 3 vs TASK 4 — DIRECT COMPARISON")
print(f"{'#'*70}")

# ── Retrieve Task 3 test predictions (re-evaluate saved model) ──
print("\n  Re-evaluating Task 3 CustomCNN for direct comparison...")
try:
    from torchvision import models as tvm

    # Rebuild Task 3 architecture inline
    class ConvBlock(nn.Module):
        def __init__(self, in_ch, out_ch, kernel_size=3, pool=True):
            super().__init__()
            layers = [
                nn.Conv2d(in_ch, out_ch, kernel_size=kernel_size,
                          padding=kernel_size // 2, bias=False),
                nn.BatchNorm2d(out_ch),
                nn.ReLU(inplace=True),
            ]
            if pool:
                layers.append(nn.MaxPool2d(2, 2))
            self.block = nn.Sequential(*layers)
        def forward(self, x): return self.block(x)

    class CustomCNN(nn.Module):
        def __init__(self, num_classes=NUM_CLASSES, d1=0.5, d2=0.4):
            super().__init__()
            self.features = nn.Sequential(
                ConvBlock(3, 32), ConvBlock(32, 64),
                ConvBlock(64, 128), ConvBlock(128, 256), ConvBlock(256, 512)
            )
            self.gap = nn.AdaptiveAvgPool2d(1)
            self.classifier = nn.Sequential(
                nn.Flatten(), nn.Dropout(d1),
                nn.Linear(512, 256), nn.ReLU(inplace=True),
                nn.Dropout(d2), nn.Linear(256, num_classes)
            )
        def forward(self, x):
            return self.classifier(self.gap(self.features(x)))

    t3_model = CustomCNN().to(DEVICE)
    t3_model.load_state_dict(torch.load(f"{TASK3_DIR}/customcnn_final.pth",
                                         map_location=DEVICE,
                                         weights_only=True))

    
    _, _, y_pred_t3, y_true_t3, y_probs_t3 = evaluate(t3_model, test_loader)
    cm_task3 = confusion_matrix(y_true_t3, y_pred_t3)
    del t3_model
    torch.cuda.empty_cache()
    print("  ✅ Task 3 re-evaluation complete.")

except Exception as e:
    print(f"  ⚠ Could not load Task 3 model: {e}")
    print("    Using zero arrays as placeholders — "
          "run Task 3 first in the same session.")
    y_pred_t3  = np.zeros_like(y_pred_t4)
    y_true_t3  = y_true_t4.copy()
    y_probs_t3 = y_probs_t4.copy()
    cm_task3   = confusion_matrix(y_true_t3, y_pred_t3)

# ── SECTION 9 extended: Side-by-side + diff confusion matrix ──
print(f"\n  SECTION 9 (extended) — CONFUSION MATRIX DIFF: Task 3 vs Task 4")
plot_confusion_matrix_diff(cm_task3, cm_task4)

# ── Per-class F1 improvement chart ──
print(f"\n  SECTION 8 (extended) — PER-CLASS F1 IMPROVEMENT")
plot_per_class_f1_comparison(y_true_t3, y_pred_t3,
                              y_true_t4, y_pred_t4)

# ── SECTION 10 extended: Overlaid ROC ──
print(f"\n  SECTION 10 (extended) — OVERLAID ROC: Task 3 vs Task 4")
plot_roc_overlay_task3_vs_task4(y_true_t3, y_probs_t3,
                                 y_true_t4, y_probs_t4)

# ── Final summary table ──
print(f"\n{'='*60}")
print(f"  FINAL SUMMARY — TASK 3 vs TASK 4")
print(f"{'='*60}")

t3_acc  = accuracy_score(y_true_t3, y_pred_t3)
t3_prec = precision_score(y_true_t3, y_pred_t3, average='macro', zero_division=0)
t3_rec  = recall_score(y_true_t3, y_pred_t3,    average='macro', zero_division=0)
t3_f1   = f1_score(y_true_t3, y_pred_t3,        average='macro', zero_division=0)

comparison_df = pd.DataFrame([
    {
        "Model"         : "CustomCNN (Task 3)",
        "Test Accuracy" : f"{t3_acc:.4f}",
        "Precision"     : f"{t3_prec:.4f}",
        "Recall"        : f"{t3_rec:.4f}",
        "F1-Score"      : f"{t3_f1:.4f}",
        "Macro AUC"     : "see Task 3",
        "KFold Mean"    : f"{np.mean(custom_folds):.4f}",
    },
    {
        "Model"         : "AttentionCNN (Task 4)",
        "Test Accuracy" : f"{acc:.4f}",
        "Precision"     : f"{prec:.4f}",
        "Recall"        : f"{rec:.4f}",
        "F1-Score"      : f"{f1:.4f}",
        "Macro AUC"     : f"{macro_auc:.4f}",
        "KFold Mean"    : f"{np.mean(attention_fold_accuracies):.4f}",
    },
])

print(comparison_df.to_string(index=False))
comparison_df.to_csv("task3_vs_task4_comparison.csv", index=False)
print("\n  ✅ Saved: task3_vs_task4_comparison.csv")
print("\n✅ Task 4 Complete.")
```

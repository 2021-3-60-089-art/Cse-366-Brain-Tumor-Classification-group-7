```python

import os
import time
import random
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import seaborn as sns
from pathlib import Path
from collections import Counter

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, Dataset, Subset
from torchvision import transforms, models
from torchvision.datasets import ImageFolder

from sklearn.model_selection import StratifiedKFold, train_test_split
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    classification_report, confusion_matrix,
    roc_auc_score, roc_curve, auc
)
from sklearn.preprocessing import label_binarize
from scipy.stats import ttest_rel

warnings.filterwarnings("ignore")

# ── Reproducibility seed (MUST stay fixed for fair comparison) ──
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"✅ Device: {DEVICE}")
print(f"✅ PyTorch version: {torch.__version__}")


# ============================================================
# SECTION 2: DATA LOADING AND PREPROCESSING
# ============================================================

# ── Shared hyperparameters (identical across all 3 models) ──
IMG_SIZE    = 224
BATCH_SIZE  = 32
NUM_EPOCHS  = 20       # Phase 1: frozen base
FINETUNE_EPOCHS = 10   # Phase 2: unfrozen last blocks
LR          = 1e-3
FINETUNE_LR = 1e-4
N_FOLDS     = 5
NUM_WORKERS = 2

DATA_DIR = Path("/kaggle/input/datasets/fernando2rad/brain-tumor-mri-images-44c")

# /kaggle/input/datasets/fernando2rad/brain-tumor-mri-images-44c


# ── Transforms ──
train_transforms = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(15),
    transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225]),
])

val_test_transforms = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225]),
])

# ── Load full dataset to extract labels for stratified splitting ──
full_dataset = ImageFolder(root=DATA_DIR)
CLASS_NAMES  = full_dataset.classes
NUM_CLASSES  = len(CLASS_NAMES)
all_labels   = np.array([label for _, label in full_dataset.samples])
all_indices  = np.arange(len(full_dataset))

print(f"✅ Total images   : {len(full_dataset)}")
print(f"✅ Number of classes: {NUM_CLASSES}")

# ── Stratified 70 / 15 / 15 split (fixed once for all models) ──
train_idx, temp_idx, train_labels, temp_labels = train_test_split(
    all_indices, all_labels,
    test_size=0.30,
    stratify=all_labels,
    random_state=SEED
)
val_idx, test_idx, _, _ = train_test_split(
    temp_idx, temp_labels,
    test_size=0.50,
    stratify=temp_labels,
    random_state=SEED
)

print(f"✅ Train size : {len(train_idx)}")
print(f"✅ Val   size : {len(val_idx)}")
print(f"✅ Test  size : {len(test_idx)}")

# ── Build dataset subsets with correct transforms ──
def make_subset(dataset, indices, transform):
    subset = Subset(dataset, indices)
    subset.dataset = ImageFolder(root=DATA_DIR, transform=transform)
    return Subset(subset.dataset, indices)

train_dataset = make_subset(full_dataset, train_idx, train_transforms)
val_dataset   = make_subset(full_dataset, val_idx,   val_test_transforms)
test_dataset  = make_subset(full_dataset, test_idx,  val_test_transforms)

val_loader  = DataLoader(val_dataset,  batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS)

# ── 5-Fold Stratified K-Fold on training indices only ──
skf = StratifiedKFold(n_splits=N_FOLDS, shuffle=True, random_state=SEED)
train_labels_for_fold = all_labels[train_idx]
fold_splits = list(skf.split(train_idx, train_labels_for_fold))

print(f"✅ K-Fold splits generated: {N_FOLDS} folds")

# ── Class distribution plot ──
plt.figure(figsize=(20, 4))
plt.title("Class Distribution Across All 44 Classes", fontsize=14, fontweight='bold')
class_counts = Counter(all_labels)
plt.bar(range(NUM_CLASSES), [class_counts[i] for i in range(NUM_CLASSES)])
plt.xticks(range(NUM_CLASSES), CLASS_NAMES, rotation=90, fontsize=6)
plt.xlabel("Class")
plt.ylabel("Count")
plt.tight_layout()
plt.savefig("class_distribution.png", dpi=150)
plt.show()


# ============================================================
# SECTION 3 — SHARED UTILITIES (used by all 3 models)
# ============================================================

def set_seed(seed=SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)


def freeze_base(model, model_name):
    """Freeze all layers except the classifier head."""
    if model_name == "mobilenetv3":
        for param in model.features.parameters():
            param.requires_grad = False
    elif model_name == "resnet50":
        for name, param in model.named_parameters():
            if "fc" not in name:
                param.requires_grad = False
    elif model_name == "efficientnetv2":
        for param in model.features.parameters():
            param.requires_grad = False


def unfreeze_last_blocks(model, model_name):
    """Unfreeze the last few blocks for fine-tuning."""
    if model_name == "mobilenetv3":
        # Unfreeze last 3 feature blocks
        for param in model.features[-3:].parameters():
            param.requires_grad = True
    elif model_name == "resnet50":
        # Unfreeze layer3 and layer4
        for name, param in model.named_parameters():
            if "layer3" in name or "layer4" in name or "fc" in name:
                param.requires_grad = True
    elif model_name == "efficientnetv2":
        # Unfreeze last 3 feature blocks
        for param in model.features[-3:].parameters():
            param.requires_grad = True


def train_one_epoch(model, loader, criterion, optimizer):
    model.train()
    running_loss, correct, total = 0.0, 0, 0
    for inputs, labels in loader:
        inputs, labels = inputs.to(DEVICE), labels.to(DEVICE)
        optimizer.zero_grad()
        outputs = model(inputs)
        loss    = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * inputs.size(0)
        _, preds = torch.max(outputs, 1)
        correct  += (preds == labels).sum().item()
        total    += labels.size(0)
    return running_loss / total, correct / total


def evaluate(model, loader):
    model.eval()
    running_loss, correct, total = 0.0, 0, 0
    criterion = nn.CrossEntropyLoss()
    all_preds, all_labels, all_probs = [], [], []
    with torch.no_grad():
        for inputs, labels in loader:
            inputs, labels = inputs.to(DEVICE), labels.to(DEVICE)
            outputs = model(inputs)
            loss    = criterion(outputs, labels)
            running_loss += loss.item() * inputs.size(0)
            probs = torch.softmax(outputs, dim=1)
            _, preds = torch.max(outputs, 1)
            correct  += (preds == labels).sum().item()
            total    += labels.size(0)
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())
            all_probs.extend(probs.cpu().numpy())
    return (running_loss / total, correct / total,
            np.array(all_preds), np.array(all_labels), np.array(all_probs))


# ── SECTION 4 (shared): TRAINING LOOP WITH TIME TRACKING ──
def run_training(model, model_name, fold=None):
    """
    Two-phase training:
      Phase 1 — frozen base, train head only
      Phase 2 — unfreeze last blocks, fine-tune
    Returns history dict and total training time.
    """
    criterion  = nn.CrossEntropyLoss()
    history    = {"train_loss": [], "train_acc": [],
                  "val_loss":   [], "val_acc":   []}
    total_time = 0.0
    tag = f"[{model_name.upper()}{'  Fold ' + str(fold) if fold else ''}]"

    # ── Phase 1: Train head only ──
    optimizer = optim.Adam(filter(lambda p: p.requires_grad, model.parameters()), lr=LR)
    scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=7, gamma=0.1)

    print(f"\n{tag} Phase 1 — Training classifier head ({NUM_EPOCHS} epochs)")
    for epoch in range(1, NUM_EPOCHS + 1):
        t0 = time.time()
        tr_loss, tr_acc = train_one_epoch(model, fold_train_loader, criterion, optimizer)
        vl_loss, vl_acc, _, _, _ = evaluate(model, fold_val_loader)
        scheduler.step()
        elapsed = time.time() - t0
        total_time += elapsed
        history["train_loss"].append(tr_loss)
        history["train_acc"].append(tr_acc)
        history["val_loss"].append(vl_loss)
        history["val_acc"].append(vl_acc)
        if epoch % 5 == 0 or epoch == 1:
            print(f"  Epoch {epoch:02d}/{NUM_EPOCHS} | "
                  f"Train Loss: {tr_loss:.4f} Acc: {tr_acc:.4f} | "
                  f"Val Loss: {vl_loss:.4f} Acc: {vl_acc:.4f} | "
                  f"Time: {elapsed:.1f}s")

    # ── Phase 2: Fine-tune last blocks ──
    unfreeze_last_blocks(model, model_name)
    optimizer = optim.Adam(filter(lambda p: p.requires_grad, model.parameters()), lr=FINETUNE_LR)
    scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=FINETUNE_EPOCHS)

    print(f"\n{tag} Phase 2 — Fine-tuning last blocks ({FINETUNE_EPOCHS} epochs)")
    for epoch in range(1, FINETUNE_EPOCHS + 1):
        t0 = time.time()
        tr_loss, tr_acc = train_one_epoch(model, fold_train_loader, criterion, optimizer)
        vl_loss, vl_acc, _, _, _ = evaluate(model, fold_val_loader)
        scheduler.step()
        elapsed = time.time() - t0
        total_time += elapsed
        history["train_loss"].append(tr_loss)
        history["train_acc"].append(tr_acc)
        history["val_loss"].append(vl_loss)
        history["val_acc"].append(vl_acc)
        if epoch % 5 == 0 or epoch == 1:
            print(f"  Epoch {epoch:02d}/{FINETUNE_EPOCHS} | "
                  f"Train Loss: {tr_loss:.4f} Acc: {tr_acc:.4f} | "
                  f"Val Loss: {vl_loss:.4f} Acc: {vl_acc:.4f} | "
                  f"Time: {elapsed:.1f}s")

    print(f"\n{tag} Total training time: {total_time:.1f}s ({total_time/60:.1f} min)")
    return history, total_time


# ── SECTION 5 (shared): PLOT TRAINING AND VALIDATION CURVES ──
def plot_curves(history, model_name, fold=None):
    tag   = f"{model_name.upper()}" + (f" — Fold {fold}" if fold else " — Full")
    total = len(history["train_loss"])
    epochs = range(1, total + 1)
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    fig.suptitle(f"Training & Validation Curves — {tag}", fontsize=14, fontweight='bold')

    axes[0].plot(epochs, history["train_loss"], label="Train Loss", color="royalblue")
    axes[0].plot(epochs, history["val_loss"],   label="Val Loss",   color="tomato")
    axes[0].axvline(x=NUM_EPOCHS, color='gray', linestyle='--', label='Fine-tune start')
    axes[0].set_title("Loss")
    axes[0].set_xlabel("Epoch")
    axes[0].set_ylabel("Loss")
    axes[0].legend()

    axes[1].plot(epochs, history["train_acc"], label="Train Acc", color="royalblue")
    axes[1].plot(epochs, history["val_acc"],   label="Val Acc",   color="tomato")
    axes[1].axvline(x=NUM_EPOCHS, color='gray', linestyle='--', label='Fine-tune start')
    axes[1].set_title("Accuracy")
    axes[1].set_xlabel("Epoch")
    axes[1].set_ylabel("Accuracy")
    axes[1].legend()

    plt.tight_layout()
    plt.savefig(f"curves_{model_name}_fold{fold}.png", dpi=150)
    plt.show()


# ── SECTION 8 (shared): CLASSIFICATION REPORT ──
def print_classification_report(y_true, y_pred, model_name):
    print(f"\n{'='*60}")
    print(f"  CLASSIFICATION REPORT — {model_name.upper()}")
    print(f"{'='*60}")
    print(classification_report(y_true, y_pred,
                                 target_names=CLASS_NAMES,
                                 digits=4))


# ── SECTION 9 (shared): CONFUSION MATRIX ──
def plot_confusion_matrix(y_true, y_pred, model_name):
    cm = confusion_matrix(y_true, y_pred)
    fig, ax = plt.subplots(figsize=(28, 26))
    ax.set_title(f"Confusion Matrix — {model_name.upper()}", fontsize=16, fontweight='bold')
    sns.heatmap(cm, annot=False, fmt='d', cmap='Blues',
                xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES, ax=ax)
    ax.set_xlabel("Predicted Label", fontsize=12)
    ax.set_ylabel("True Label", fontsize=12)
    plt.xticks(rotation=90, fontsize=6)
    plt.yticks(rotation=0,  fontsize=6)
    plt.tight_layout()
    plt.savefig(f"confusion_matrix_{model_name}.png", dpi=150)
    plt.show()


# ── SECTION 10 (shared): ROC CURVE AND AUC (ONE VS REST) ──
def plot_roc_curves(y_true, y_probs, model_name):
    y_bin = label_binarize(y_true, classes=list(range(NUM_CLASSES)))
    fpr, tpr, roc_auc_dict = {}, {}, {}

    for i in range(NUM_CLASSES):
        fpr[i], tpr[i], _ = roc_curve(y_bin[:, i], y_probs[:, i])
        roc_auc_dict[i]   = auc(fpr[i], tpr[i])

    # Macro average
    all_fpr  = np.unique(np.concatenate([fpr[i] for i in range(NUM_CLASSES)]))
    mean_tpr = np.zeros_like(all_fpr)
    for i in range(NUM_CLASSES):
        mean_tpr += np.interp(all_fpr, fpr[i], tpr[i])
    mean_tpr /= NUM_CLASSES
    macro_auc = auc(all_fpr, mean_tpr)

    plt.figure(figsize=(10, 8))
    plt.title(f"ROC Curves (One-vs-Rest) — {model_name.upper()}", fontsize=14, fontweight='bold')
    for i in range(NUM_CLASSES):
        plt.plot(fpr[i], tpr[i], alpha=0.3, linewidth=0.8)
    plt.plot(all_fpr, mean_tpr, color='navy', linewidth=2.5,
             label=f"Macro Average AUC = {macro_auc:.4f}")
    plt.plot([0, 1], [0, 1], 'k--', linewidth=1)
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.legend(loc="lower right")
    plt.tight_layout()
    plt.savefig(f"roc_{model_name}.png", dpi=150)
    plt.show()

    return macro_auc


# ── SECTION 7 (shared): PERFORMANCE METRICS ──
def report_metrics(y_true, y_pred, y_probs, model_name, training_time):
    acc  = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred, average='macro', zero_division=0)
    rec  = recall_score(y_true, y_pred, average='macro', zero_division=0)
    f1   = f1_score(y_true, y_pred, average='macro', zero_division=0)

    print(f"\n{'='*60}")
    print(f"  PERFORMANCE METRICS — {model_name.upper()}")
    print(f"{'='*60}")
    print(f"  Accuracy          : {acc:.4f}")
    print(f"  Precision (macro) : {prec:.4f}")
    print(f"  Recall (macro)    : {rec:.4f}")
    print(f"  F1-Score (macro)  : {f1:.4f}")
    print(f"  Training Time     : {training_time:.1f}s ({training_time/60:.1f} min)")
    return acc, prec, rec, f1


# ============================================================
# MAIN K-FOLD LOOP — runs for each model
# ============================================================

def run_model_pipeline(model_name, build_model_fn):
    global fold_train_loader, fold_val_loader
    """
    Full pipeline for one model:
      - 5-Fold K-Fold on training set for tuning
      - Final train on full 70% train set
      - Evaluate on held-out 15% test set
      - All 12 steps
    """
    print(f"\n{'#'*70}")
    print(f"#  STARTING PIPELINE FOR: {model_name.upper()}")
    print(f"{'#'*70}")

    fold_accuracies = []
    fold_histories  = []

    # ── SECTION 4: 5-FOLD K-FOLD TRAINING ──
    print(f"\n{'='*60}")
    print(f"  SECTION 4 — 5-FOLD STRATIFIED K-FOLD TRAINING [{model_name.upper()}]")
    print(f"{'='*60}")

    for fold_idx, (fold_train_rel, fold_val_rel) in enumerate(fold_splits, 1):
        print(f"\n── FOLD {fold_idx}/{N_FOLDS} ──")
        set_seed(SEED)

        # Map relative fold indices back to absolute dataset indices
        fold_train_abs = train_idx[fold_train_rel]
        fold_val_abs   = train_idx[fold_val_rel]

        # Build fold datasets with correct transforms
        fold_train_ds = Subset(
            ImageFolder(root=DATA_DIR, transform=train_transforms),
            fold_train_abs
        )
        fold_val_ds = Subset(
            ImageFolder(root=DATA_DIR, transform=val_test_transforms),
            fold_val_abs
        )

        
        fold_train_loader = DataLoader(fold_train_ds, batch_size=BATCH_SIZE,
                                        shuffle=True,  num_workers=NUM_WORKERS)
        fold_val_loader   = DataLoader(fold_val_ds,   batch_size=BATCH_SIZE,
                                        shuffle=False, num_workers=NUM_WORKERS)

        model = build_model_fn().to(DEVICE)
        freeze_base(model, model_name)

        history, _ = run_training(model, model_name, fold=fold_idx)
        fold_histories.append(history)

        # Evaluate fold on its val portion
        _, fold_acc, _, _, _ = evaluate(model, fold_val_loader)
        fold_accuracies.append(fold_acc)
        print(f"  Fold {fold_idx} Val Accuracy: {fold_acc:.4f}")

        # ── SECTION 5: Plot fold curves ──
        print(f"\n  SECTION 5 — Training & Validation Curves [Fold {fold_idx}]")
        plot_curves(history, model_name, fold=fold_idx)

        del model
        torch.cuda.empty_cache()

    print(f"\n── K-Fold Summary [{model_name.upper()}] ──")
    print(f"  Per-fold accuracies : {[f'{a:.4f}' for a in fold_accuracies]}")
    print(f"  Mean accuracy       : {np.mean(fold_accuracies):.4f} ± {np.std(fold_accuracies):.4f}")

    # ── FINAL MODEL: train on full 70% train set, evaluate on 15% test set ──
    print(f"\n{'='*60}")
    print(f"  FINAL MODEL TRAINING ON FULL TRAINING SET [{model_name.upper()}]")
    print(f"{'='*60}")
    set_seed(SEED)

    
    fold_train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE,
                                    shuffle=True,  num_workers=NUM_WORKERS)
    fold_val_loader   = val_loader

    final_model = build_model_fn().to(DEVICE)
    if torch.cuda.device_count() > 1:
        final_model = nn.DataParallel(final_model)
    freeze_base(final_model, model_name)

    final_history, total_time = run_training(final_model, model_name, fold=None)

    # ── SECTION 5: Final model curves ──
    print(f"\n  SECTION 5 — Training & Validation Curves [Final Model]")
    plot_curves(final_history, model_name, fold="final")

    # ── SECTIONS 6 & 7: Evaluation and performance metrics ──
    print(f"\n  SECTION 6 & 7 — EVALUATION AND PERFORMANCE METRICS [{model_name.upper()}]")
    _, test_acc, y_pred, y_true, y_probs = evaluate(final_model, test_loader)

    acc, prec, rec, f1 = report_metrics(y_true, y_pred, y_probs, model_name, total_time)

    # ── SECTION 8: Classification report ──
    print(f"\n  SECTION 8 — CLASSIFICATION REPORT [{model_name.upper()}]")
    print_classification_report(y_true, y_pred, model_name)

    # ── SECTION 9: Confusion matrix ──
    print(f"\n  SECTION 9 — CONFUSION MATRIX [{model_name.upper()}]")
    plot_confusion_matrix(y_true, y_pred, model_name)

    # ── SECTION 10: ROC Curve and AUC ──
    print(f"\n  SECTION 10 — ROC CURVE AND AUC (ONE VS REST) [{model_name.upper()}]")
    macro_auc = plot_roc_curves(y_true, y_probs, model_name)

    # ── SECTION 12: Final AUC score ──
    print(f"\n  SECTION 12 — FINAL DETECTION AUC SCORE [{model_name.upper()}]")
    print(f"  Macro-Average AUC (One-vs-Rest): {macro_auc:.4f}")

    # Save model
    torch.save(final_model.state_dict(), f"{model_name}_final.pth")
    print(f"\n  ✅ Model saved: {model_name}_final.pth")

    return {
        "model_name"       : model_name,
        "fold_accuracies"  : fold_accuracies,
        "test_accuracy"    : acc,
        "precision"        : prec,
        "recall"           : rec,
        "f1"               : f1,
        "macro_auc"        : macro_auc,
        "training_time_s"  : total_time,
    }


# ============================================================
# SECTION 3A: MODEL ARCHITECTURE — MobileNetV3
# ============================================================

def build_mobilenetv3():
    print("\n── Building MobileNetV3-Large ──")
    model = models.mobilenet_v3_large(weights=models.MobileNet_V3_Large_Weights.DEFAULT)
    in_features = model.classifier[3].in_features
    model.classifier[3] = nn.Linear(in_features, NUM_CLASSES)
    return model

results_mobilenet = run_model_pipeline("mobilenetv3", build_mobilenetv3)


# ============================================================
# SECTION 3B: MODEL ARCHITECTURE — ResNet50
# ============================================================

def build_resnet50():
    print("\n── Building ResNet50 ──")
    model = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)
    in_features = model.fc.in_features
    model.fc = nn.Linear(in_features, NUM_CLASSES)
    return model

results_resnet = run_model_pipeline("resnet50", build_resnet50)


# ============================================================
# SECTION 3C: MODEL ARCHITECTURE — EfficientNetV2
# ============================================================

def build_efficientnetv2():
    print("\n── Building EfficientNetV2-S ──")
    model = models.efficientnet_v2_s(weights=models.EfficientNet_V2_S_Weights.DEFAULT)
    in_features = model.classifier[1].in_features
    model.classifier[1] = nn.Linear(in_features, NUM_CLASSES)
    return model

results_efficientnet = run_model_pipeline("efficientnetv2", build_efficientnetv2)


# ============================================================
# SECTION 11: PAIRED T-TEST BETWEEN MODEL ACCURACY SCORES
# ============================================================

print(f"\n{'='*60}")
print(f"  SECTION 11 — PAIRED T-TEST ACROSS MODELS")
print(f"{'='*60}")

mob_folds = results_mobilenet["fold_accuracies"]
res_folds = results_resnet["fold_accuracies"]
eff_folds = results_efficientnet["fold_accuracies"]

t_mob_res, p_mob_res = ttest_rel(mob_folds, res_folds)
t_mob_eff, p_mob_eff = ttest_rel(mob_folds, eff_folds)
t_res_eff, p_res_eff = ttest_rel(res_folds, eff_folds)

print(f"\n  MobileNetV3 vs ResNet50     : t={t_mob_res:.4f}, p={p_mob_res:.4f} "
      f"{'→ Significant ✅' if p_mob_res < 0.05 else '→ Not significant ❌'}")
print(f"  MobileNetV3 vs EfficientNetV2: t={t_mob_eff:.4f}, p={p_mob_eff:.4f} "
      f"{'→ Significant ✅' if p_mob_eff < 0.05 else '→ Not significant ❌'}")
print(f"  ResNet50 vs EfficientNetV2   : t={t_res_eff:.4f}, p={p_res_eff:.4f} "
      f"{'→ Significant ✅' if p_res_eff < 0.05 else '→ Not significant ❌'}")


# ============================================================
# FINAL CROSS-MODEL COMPARISON SUMMARY
# ============================================================

print(f"\n{'='*60}")
print(f"  FINAL CROSS-MODEL COMPARISON SUMMARY")
print(f"{'='*60}")

all_results = [results_mobilenet, results_resnet, results_efficientnet]
summary_df  = pd.DataFrame([{
    "Model"         : r["model_name"].upper(),
    "Test Accuracy" : f"{r['test_accuracy']:.4f}",
    "Precision"     : f"{r['precision']:.4f}",
    "Recall"        : f"{r['recall']:.4f}",
    "F1-Score"      : f"{r['f1']:.4f}",
    "Macro AUC"     : f"{r['macro_auc']:.4f}",
    "Train Time (s)": f"{r['training_time_s']:.1f}",
    "KFold Mean Acc": f"{np.mean(r['fold_accuracies']):.4f}",
    "KFold Std Acc" : f"{np.std(r['fold_accuracies']):.4f}",
} for r in all_results])

print(summary_df.to_string(index=False))

# ── Bar chart comparison ──
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle("Cross-Model Comparison", fontsize=16, fontweight='bold')

model_labels = [r["model_name"].upper() for r in all_results]
colors       = ["steelblue", "darkorange", "seagreen"]

metrics = [
    ("test_accuracy", "Test Accuracy",  axes[0]),
    ("f1",            "Macro F1-Score", axes[1]),
    ("macro_auc",     "Macro AUC",      axes[2]),
]
for key, title, ax in metrics:
    vals = [r[key] for r in all_results]
    bars = ax.bar(model_labels, vals, color=colors)
    ax.set_title(title, fontweight='bold')
    ax.set_ylim(0, 1.05)
    for bar, val in zip(bars, vals):
        ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01,
                f"{val:.4f}", ha='center', va='bottom', fontsize=11)

plt.tight_layout()
plt.savefig("cross_model_comparison.png", dpi=150)
plt.show()

# ── Overlaid ROC curves (all 3 models, macro average only) ──
print("\n── Overlaid Macro-Average ROC Curves (All 3 Models) ──")
# NOTE: To overlay, re-evaluate each saved model here
# This plots AUC scores as a bar for a clean final view
plt.figure(figsize=(8, 6))
plt.title("Macro-Average AUC Comparison", fontsize=14, fontweight='bold')
auc_vals = [r["macro_auc"] for r in all_results]
bars = plt.bar(model_labels, auc_vals, color=colors, edgecolor='black')
for bar, val in zip(bars, auc_vals):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.005,
             f"{val:.4f}", ha='center', va='bottom', fontsize=12, fontweight='bold')
plt.ylim(0, 1.1)
plt.ylabel("Macro AUC (One-vs-Rest)")
plt.tight_layout()
plt.savefig("auc_comparison.png", dpi=150)
plt.show()

print("\n✅ All sections complete. task2 finished.")
```
